```python
# Game name: Can You Clean the Water?
# Hong Kong Water Treatment Simulator- A Python-based environmental engineering quiz and decision game

dic_questions = {
    "River water after typhoon season": "sedimentation",

    "Water with harmful bacteria": "chlorination",

    "Seawater used for toilet flushing in Hong Kong": "screening and disinfection",

    "Water containing large suspended particles": "filtration",

    "Salt water that needs to become drinking water": "desalination",

    "Water with bad odor and organic pollutants": "activated carbon",

    "Wastewater with microorganisms and organic waste": "biological treatment",

    "Water containing heavy metals": "chemical precipitation",

    "Algae bloom in reservoir water": "activated carbon",

    "Microplastics in wastewater": "membrane filtration"
}


#random module is imported to randomly pick questions from the dictionary
import random
def trivia_game():
  score=0
  # Convert the dictionary keys into a list to allow iteration over them
  list_questions=list(dic_questions.keys())
  # This picks two random questions from the list
  selected_questions=random.sample(list_questions,4)
  # allows questions to be displayed with index
  # 1 is added to the index as I want the index to start from 1
  for idx,question in enumerate(selected_questions):
    print(f"{idx+1}.{question}")
    # As the answers are in lowercase I used "lower.()"
    # In case the user enters a space I used strip
    answer=input("Enter your anwer: ").lower().strip()
    correct_answer=dic_questions[question]

    if correct_answer==answer:
      print(f"Good job! Your answer is correct.")
      score+=1
    else:
      print(f"Your answer is incorrect.The correct answer is {correct_answer}")

  print(f"Game over!. Your final score is {score}/4")

trivia_game()
```
