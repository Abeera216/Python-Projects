```python
#Plover Cove Reservoir Monitoring System

import random

def contamination_detector():
  print("Scenario: Recent typhoon activity has increased runoff into the reservoir.")
  print("Environmental engineers suspect possible contamination.")
  correct_answer=random.randint(1,101)
  attempt=0
  while True:
    user_guess=int(input("Estimate the contamination level between 1-100."))
    attempt+=1
    if correct_answer == user_guess:
      print(f"Good job! It took {attempt} attempts")
      break
    elif correct_answer in range(1,21):
      print("Incorrect anwer. Hint: Water appears clear. This suggests contamination levels between 1-20.")
    elif correct_answer in range(21,41):
      print("Incorrect anwer. Hint: Slight turbidity detected.This suggests contamination levels between 21-40.")
    elif correct_answer in range(41,60):
      print("Incorrect anwer. Hint: Increased algae growth observed. This suggests contamination levels between 41-60.")
    elif correct_answer in range(61,81):
      print("Incorrect anwer. Hint: Fish stress detected. This suggests contamination levels between 61-80.")
    elif correct_answer in range(81,101):
      print("Incorrect anwer. Hint: Public health warning issued. This suggests contamination levels between 81-100.")



contamination_detector()


```
