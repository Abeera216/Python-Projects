```python
# Password generator
import random
import string

def password_generator():
  # This password is program to always have lowercase letters.
  length=int(input("What is the length of the password?"))
  to_include_upper=input("Include uppercase letters or not? (yes/no)").lower().strip()
  to_include_special=input("Include special characters or not? (yes/no)").lower().strip()
  to_include_digits=input("Include digits or not? (yes/no)").lower().strip()

  if length < 4:
    print("Invalid lengh of the password.The password must atleast be 4 characters long")
    return


  lower=string.ascii_lowercase
  upper=string.ascii_uppercase if to_include_upper=="yes" else ""
  digits=string.digits if to_include_upper=="yes" else ""
  special=string.punctuation if to_include_upper=="yes" else ""
  all_character=lower+upper+digits+special
  required_characters=[]

  if to_include_upper=="yes":
    required_characters.append(random.choice(upper))
  if to_include_digits=="yes":
    required_characters.append(random.choice(digits))
  if to_include_special=="yes":
    required_characters.append(random.choice(special))

  length_to_be_randomly_filled=length-len(required_characters)

  for _ in range(length_to_be_randomly_filled):
    randomly_filled=random.sample(all_character,length_to_be_randomly_filled)
    password_list=required_characters+ randomly_filled
    random.shuffle(password_list)
  password="".join(password_list)
  print(f"Your randomly generated password is {password}")




password_generator()



```
