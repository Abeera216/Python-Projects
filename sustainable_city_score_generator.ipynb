```python
# Sustainable City Score Generator
import random

def sustainable_city_score_generator():
  to_include_transport_systems=input("Include transport_systems or not? (yes/no)").lower().strip()
  to_include_waste_systems=input("Include waste_systems or not? (yes/no)").lower().strip()
  to_include_energy_types=input("Include energy_types or not? (yes/no)").lower().strip()


  energy_types = {
    "solar": 30,
    "wind": 28,
    "hydropower": 25,
    "natural gas": 10,
    "coal": -20
   }
  transport_systems = {
    "electric metro": 30,
    "bike-friendly roads": 25,
    "electric buses": 20,
    "diesel buses": -10,
    "car-dependent highways": -25
   }
  waste_systems = {
    "recycling": 30,
    "composting": 25,
    "waste-to-energy": 15,
    "incineration": 5,
    "landfills": -20
  }

  score=0

  if to_include_transport_systems == "yes":
    # Randomly pick a transport system
    transport_choice = random.choice(list(transport_systems.keys()))
    transport_score = transport_systems[transport_choice]
    score += transport_score
    print(f"Selected transport: {transport_choice} (+{transport_score} points)")

  if to_include_waste_systems == "yes":
    # Randomly pick a waste system
    waste_choice = random.choice(list(waste_systems.keys()))
    waste_score = waste_systems[waste_choice]
    score += waste_score
    print(f"Selected waste system: {waste_choice} (+{waste_score} points)")

  if to_include_energy_types == "yes":
    # Randomly pick an energy type
    energy_choice = random.choice(list(energy_types.keys()))
    energy_score = energy_types[energy_choice]
    score += energy_score
    print(f"Selected energy type: {energy_choice} (+{energy_score} points)")

  print(f"The Sustainable City Score Generator is {score}")




sustainable_city_score_generator()
```
