```python
import random
def guess_the_number():
  attempts=0
  correct_answer=random.randrange(1,11)
  while True:
    # Since the input needs to be continously asked it goes in the while loop
    user_answer=int(input("Guess the number between 1 and 10 inclusive: "))
    # The number of attempts should increase every time the user makes a guess
    attempts+=1
    if user_answer==correct_answer:
      print(f"Correct answer! It took {attempts} attempt")
      break
    # elif requires a condition
    elif user_answer!=correct_answer:
      print("Try Again!")

guess_the_number()

```
